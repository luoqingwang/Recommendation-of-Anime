```python
import pandas as pd

animes = pd.read_csv("data/animes.csv")
animes = animes.drop(columns=[animes.columns[0]])
new_title_column = animes.columns[0]
animes.columns = ['title'] + list(animes.columns[1:])
animes = animes.drop_duplicates(subset=new_title_column, keep='first')
animes = animes.dropna()
animes
```

```python
animes.head()
```

```python
import matplotlib.pyplot as plt

values = ['episodes', 'members', 'popularity', 'ranked', 'score']

for i in values:
    plt.hist(animes[i])
    plt.title(i)
    plt.show()
```

```python
import seaborn as sns

#Slice out personality scores and get correlation
correlation = animes[values].corr()

#Heatmap of correlation
sns.heatmap(correlation, annot = True, cmap="Blues")
```
